# Sparse Search - TF-IDF
### Implement

```python
import re
import numpy as np
```

```python
def tokenize(text: str):
    text = text.lower()
    return re.findall(r"[a-z0-9]+", text)
```

```python
def computeTF(token: str, doc: str):
  token = tokenize(token)[0]
  tokened_doc = tokenize(doc)
  size_d = 0
  count_t_d = 0
  if len(tokened_doc) == 0:
    return 0.0
  for word in tokened_doc:
    if word == token:
      count_t_d += 1
    size_d += 1
  return count_t_d / size_d    
```

```python
def computeDF(token: str, corpus: list):
  DF = 0
  token = tokenize(token)[0]
  for doc in corpus:
    tokened_doc = tokenize(doc)
    if token in tokened_doc:
      DF += 1
  return DF
```

```python
def computeIDF(token: str, corpus: list):
  size_c = len(corpus)
  DF = computeDF(token, corpus)
  return np.log(size_c / (1 + DF))
```

```python
def computeTFIDF(token: str, doc: str, corpus: list):
  return computeTF(token, doc) * computeIDF(token, corpus)
```

```python
def sparse_search(query: str, corpus: list, num_results_to_return: int=3):
  tokened_query = tokenize(query)
  tfidf_scores = {}
  for token in tokened_query:
    for doc in corpus:
      curr_tfidf = computeTFIDF(token, doc, corpus)
      if doc not in tfidf_scores:
        tfidf_scores[doc] = curr_tfidf
      else:
        tfidf_scores[doc] += curr_tfidf
  # Sort the documents by TF-IDF score in descending order
  sorted_docs = sorted(tfidf_scores.items(), key=lambda x: x[1], reverse=True)
  # Return the top num_results_to_return documents
  return sorted_docs[:num_results_to_return]
  
```

# Dense Search

### Implement

```python
from sentence_transformers import SentenceTransformer
import numpy as np
model_name = 'sentence-transformers/all-mpnet-base-v2'
```

```python
# Ingestion - run this only once, not per query
  
model = SentenceTransformer(model_name)
embedded_corpus = model.encode(corpus, convert_to_numpy=True, show_progress_bar=True)
print(embedded_corpus.shape)
```

```python
def dense_search(query: str, embedded_corpus: np.ndarray, model: SentenceTransformer, num_results_to_return: int=3):
  query_embedding = model.encode(query, convert_to_numpy=True)
  similarities = embedded_corpus @ query_embedding
  top_docs = np.argsort(similarities)[-num_results_to_return:][::-1]
  return [(corpus[i], similarities[i]) for i in top_docs]
```

### Test your implementation
Use the Embedded corpus, and the test querys

```python
for query in test_querys:
    print(query)
    for doc, score in dense_search(query, embedded_corpus, model):
        print(doc.strip(), "-", score)
    print()
```
